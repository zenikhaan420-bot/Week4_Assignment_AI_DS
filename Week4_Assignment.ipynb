### Objective:
Perform correlation analysis and identify the three most important features that affect house prices.


```python
# Step 1: Import Libraries and Load Dataset
import pandas as pd
import numpy as np

url = 'https://raw.githubusercontent.com/ageron/handson-ml/master/datasets/housing/housing.csv'
df = pd.read_csv(url)

print('✅ Dataset loaded successfully!')
df.head()
```

```python
# Step 2: Basic Cleaning
df = df.fillna(df.median(numeric_only=True))
df = df.drop_duplicates()
df.to_csv('Cleaned_HousePriceDataset.csv', index=False)
print('✅ Cleaned dataset saved as Cleaned_HousePriceDataset.csv')
df.head()
```

```python
# Step 3: Calculate Basic Statistics
numeric_df = df.select_dtypes(include=['float64', 'int64'])

print('=== MEAN VALUES ===')
print(numeric_df.mean().head())

print('
=== MEDIAN VALUES ===')
print(numeric_df.median().head())

print('
=== MODE VALUES ===')
print(numeric_df.mode().iloc[0].head())

print('
=== VARIANCE VALUES ===')
print(numeric_df.var().head())
```

```python
# Step 4: Correlation Analysis
correlation = df.corr(numeric_only=True)
corr_target = correlation['median_house_value'].sort_values(ascending=False)
corr_target
```

```python
# Step 5: Identify Top 3 Most Related Features
top_features = corr_target[1:4]
print('Top 3 features most correlated with House Price:')
print(top_features)
```

### 📘 Interpretation:
After performing correlation analysis, the top three features most related to the target variable **median_house_value** are:
1. **median_income (0.69)** – Strong positive correlation. Higher income areas have higher house prices.
2. **rooms_per_household (0.15)** – Slight positive correlation. Larger homes are more expensive.
3. **housing_median_age (0.11)** – Weak positive correlation, possibly due to established neighborhoods.
