```python
# Cell A: recreate cleaned dataset (only if file is missing)
import pandas as pd

url = "https://raw.githubusercontent.com/ageron/handson-ml/master/datasets/housing/housing.csv"
df = pd.read_csv(url)

# Quick cleaning: fill numerical missing values with median, drop duplicates
df = df.fillna(df.median(numeric_only=True))
df = df.drop_duplicates().reset_index(drop=True)

# Save cleaned dataset to session
df.to_csv("Cleaned_HousePriceDataset.csv", index=False)
print("Saved cleaned dataset as Cleaned_HousePriceDataset.csv — rows:", df.shape[0])
```

```python
# Cell 1: load cleaned dataset and show head
import pandas as pd
df = pd.read_csv("Cleaned_HousePriceDataset.csv")
print("Dataset shape:", df.shape)
df.head()
```

```python
# Cell 2: mean, median, mode, variance (for numeric columns)
numeric = df.select_dtypes(include=['float64','int64']).copy()

means = numeric.mean()
medians = numeric.median()
modes = numeric.mode().iloc[0]           # mode() can return multiple, take first row
variances = numeric.var()

# Display summary for the target and a few important features
features_of_interest = ['median_house_value','median_income','total_rooms','housing_median_age']
print("=== Summary (selected features) ===\n")
for feat in features_of_interest:
    print(f"{feat}: mean={means[feat]:.2f}, median={medians[feat]:.2f}, mode={modes[feat]}, var={variances[feat]:.2f}")
```

```python
# Cell 3: correlation analysis
corr = df.corr(numeric_only=True)
corr_target = corr['median_house_value'].sort_values(ascending=False)
corr_target
```

```python
# Cell 4: top 3 features related to target (excluding the target itself)
top3 = corr_target.drop(labels='median_house_value').head(3)
print("Top 3 features most correlated with median_house_value:")
print(top3)
```
